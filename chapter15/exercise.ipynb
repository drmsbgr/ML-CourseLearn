```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = pd.get_dummies(df, drop_first=True, dtype=int)
df
```

```python
scaler = StandardScaler()
scaler.fit_transform(df[["Age", "RestingBP", "Cholesterol", "MaxHR", "Oldpeak"]])
df
```

```python
X = df.drop("HeartDisease", axis="columns")
y = df["HeartDisease"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
models = [SVC(), RandomForestClassifier(), LogisticRegression(max_iter=10000)]

for model in models:
    model.fit(X_train, y_train)
    print(model.score(X_test, y_test))
```

```python
pca = PCA(n_components=8)
X_pca = pca.fit_transform(X)
```

```python
X_pca_train, X_pca_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.3)
```
