```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("home-prices.csv")

df.head()
```

```python
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df["area"], df["price"])
```

```python
rg = linear_model.LinearRegression()
rg.fit(df[["area"]], df["price"])
```

```python
areas = pd.read_csv("areas.csv")

areas.head()
```

```python
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df["area"], df["price"], c="red")
plt.plot(areas, rg.predict(areas), color="blue")
```
