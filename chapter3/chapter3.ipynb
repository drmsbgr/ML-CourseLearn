# Gradient Descent

```python
import numpy as np
```

```python
def gradient_descent(x, y, lr=0.001, N=10000):
    m_curr = b_curr = 0
    n = len(x)

    for epoch in range(1, N):
        pred = m_curr * x + b_curr

        loss_deriv = pred - y

        dm = np.sum(x * loss_deriv) / n
        db = np.sum(loss_deriv) / n

        m_curr -= dm * lr
        b_curr -= db * lr

        total_loss = np.sum(0.5 / n * np.square(pred - y))
        if epoch % 10000 == 0:
            print(f"epoch: {epoch}, loss = {total_loss}")

    print(f"y={m_curr} * x + {b_curr}")


x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 7, 9, 11, 13])

gradient_descent(x, y)
```
