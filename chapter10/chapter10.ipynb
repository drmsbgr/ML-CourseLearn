```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
```

```python
iris = load_iris()
dir(iris)
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["target"] = iris.target
```

```python
X = df.drop("target", axis="columns")
y = df["target"]
```

```python
kf = StratifiedKFold(5)
```

```python
models = {
    "SVC": SVC(),
    "RFC-60": RandomForestClassifier(n_estimators=60),
    "RFC-40": RandomForestClassifier(n_estimators=40),
    "RFC-20": RandomForestClassifier(n_estimators=20),
    "RFC-10": RandomForestClassifier(n_estimators=10),
    "LogisticRegression": LogisticRegression(max_iter=500),
}
```

```python
for name, model in models.items():
    print(f"{name} {np.mean(cross_val_score(model, X, y))}")
```
